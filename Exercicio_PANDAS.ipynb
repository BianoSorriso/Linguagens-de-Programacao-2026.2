# Exercício – Análise da Qualidade do Ar no Brasil

**Notebook completo com código, respostas e interpretações.**

Os resultados abaixo foram calculados diretamente a partir do arquivo `qualidade_ar_brasil.csv`.

## Parte 1 – Preparação dos dados

```python
import pandas as pd

df = pd.read_csv('/content/qualidade_ar_brasil.csv')
df.head()
```

## Parte 2 – Exploração inicial

### Código

```python
print("Dimensões:", df.shape)
display(df.head())
df.info()
display(df.describe())
display(df.isnull().sum())
```

### Respostas – Parte 2

- **Número de medições:** 800
- **Número de cidades:** 10
- **Colunas de poluentes:** `pm25`, `pm10`, `no2` e `o3`.
- **Valores ausentes:** 0 no total.

**Interpretação:** a base possui 800 registros distribuídos entre 10 cidades. Não há valores ausentes, portanto não é necessário fazer tratamento de dados faltantes para as análises propostas.

## Parte 3 – Análise por cidade

### Código

```python
print("Cidades:")
print(df["cidade"].unique())

media_cidade = df.groupby("cidade")[["pm25", "pm10", "no2", "o3"]].mean()
display(media_cidade)

display(media_cidade.sort_values("pm25", ascending=False))
```

### Respostas – Parte 3

- **Cidade com maior média de PM2.5:** **São Paulo**, com média de **33,65**.
- **Cidade com maior média de NO2:** **São Paulo**, com média de **39,31**.
- **Cidade que apresenta níveis consistentemente altos:** **São Paulo** se destaca principalmente por PM2.5 e NO2, indicando maior pressão de poluição entre as cidades analisadas.

**Interpretação:** a comparação das médias mostra quais cidades apresentam maior concentração média dos principais poluentes.

## Parte 4 – Índice de qualidade do ar por cidade

### Código

```python
media_indice = df.groupby("cidade")["indice_qualidade_ar"].mean().sort_values(ascending=False)
display(media_indice)
```

### Respostas – Parte 4

- **Pior cidade em média:** **São Paulo**, com índice médio de **59,35**.
- **Melhor cidade em média:** **Salvador**, com índice médio de **42,34**.

**Interpretação:** quanto maior o índice de qualidade do ar utilizado na base, pior é a condição registrada. Assim, São Paulo apresenta a situação média mais preocupante, enquanto Salvador apresenta a melhor situação entre as cidades analisadas.

## Parte 5 – Situações críticas

### Código

```python
criticos = df[df["nivel_alerta"].isin(["Ruim", "Muito Ruim"])]
print("Total de registros críticos:", len(criticos))
display(criticos["cidade"].value_counts())
```

### Respostas – Parte 5

- **Total de registros classificados como “Ruim” ou “Muito Ruim”:** **14**.
- **Cidade(s) com mais registros críticos:** **São Paulo**, com **8** registro(s).

**Interpretação:** esses registros representam os episódios mais preocupantes da base e ajudam a identificar onde a qualidade do ar atingiu níveis de alerta.

## Parte 6 – Piores dias

### Código

```python
df["data"] = pd.to_datetime(df["data"])

piores_10 = df.sort_values("indice_qualidade_ar", ascending=False).head(10)
display(piores_10)
```

### Respostas – Parte 6

- **Pior registro/dia:** **19/08/2025**.
- **Cidade:** **São Paulo**.
- **Índice de qualidade do ar:** **86,30**.
- **Poluentes nesse registro, do maior para o menor:** **PM10 = 78,80**, **PM25 = 55,20**, **NO2 = 55,00**, **O3 = 45,30**.

**Interpretação:** esse registro representa o episódio de pior qualidade do ar encontrado na base. O poluente com maior valor nesse dia foi **PM10**.

## Parte 7 – Análise por mês

### Código

```python
df["mes"] = df["data"].dt.month

media_mes = df.groupby("mes")["indice_qualidade_ar"].mean().sort_values(ascending=False)
display(media_mes)
```

### Respostas – Parte 7

- **Mês com pior média de qualidade do ar:** **Junho (mês 6)**, com média de **52,88**.
- **Segundo pior mês:** **Julho**, com média de **51,29**.
- **Terceiro pior mês:** **Setembro**, com média de **49,92**.

**Interpretação:** há variação da qualidade do ar ao longo do ano. O mês de **Junho** apresentou a maior média do índice e, portanto, foi o período mais problemático na base analisada.

## Parte 8 – Relação entre temperatura e poluentes

### Código

```python
correlacao = df[["temperatura", "pm25", "pm10", "no2", "o3"]].corr()
display(correlacao)

print("Correlação da temperatura com os poluentes:")
display(correlacao["temperatura"].drop("temperatura").sort_values())
```

### Respostas – Parte 8

**Correlação da temperatura com os poluentes:**

- NO2: **-0,17**
- PM25: **-0,11**
- PM10: **-0,11**
- O3: **0,16**

- A relação mais forte entre dois poluentes foi **PM25 × PM10**, com correlação de **0,25**.
- Em geral, uma correlação próxima de **1** indica relação positiva forte; próxima de **-1**, relação negativa forte; e próxima de **0**, relação linear fraca.

**Interpretação:** a correlação mostra associação linear, mas **não prova causalidade**. Portanto, os resultados indicam relações entre as variáveis, sem permitir concluir que uma variável necessariamente cause a outra.

## Relatório final

### Código

```python
print("Resumo final")
print("Pior cidade:", df.groupby("cidade")["indice_qualidade_ar"].mean().idxmax())
print("Melhor cidade:", df.groupby("cidade")["indice_qualidade_ar"].mean().idxmin())
print("Pior mês:", df.groupby(df["data"].dt.month)["indice_qualidade_ar"].mean().idxmax())
```

### Respostas finais

**1. Qual cidade apresentou a pior qualidade do ar?**  
**São Paulo**, pois apresentou a maior média do índice de qualidade do ar.

**2. Quais poluentes foram mais críticos?**  
Os principais destaques foram **PM2.5** e **NO2**, especialmente em **São Paulo**, que apresentou as maiores médias desses poluentes entre as cidades analisadas.

**3. Quais meses foram mais problemáticos?**  
O mês mais problemático foi **Junho**, seguido por **Julho** e **Setembro**, considerando a média do índice de qualidade do ar.

**4. Quais cidades precisam de maior atenção?**  
**São Paulo** merece maior atenção por apresentar o pior índice médio e também as maiores médias de PM2.5 e NO2. As demais cidades podem ser comparadas pela média do índice e pela frequência de alertas críticos.

### Conclusão

A análise dos dados mostra diferenças importantes na qualidade do ar entre as cidades brasileiras da base. **São Paulo** apresentou a situação média mais preocupante, enquanto **Salvador** apresentou a melhor média. Os poluentes **PM2.5 e NO2** tiveram destaque nas concentrações médias, e **Junho** foi o mês com maior índice médio. Os registros classificados como **Ruim** ou **Muito Ruim** também ajudam a localizar os episódios críticos. Dessa forma, o uso conjunto das médias, dos níveis de alerta, dos piores registros e das correlações permite identificar os principais pontos de atenção para o monitoramento da qualidade do ar.
