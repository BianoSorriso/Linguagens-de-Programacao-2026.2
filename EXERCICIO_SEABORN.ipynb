## 1. Configuração do Ambiente e Base de Dados

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")

np.random.seed(101)
n = 300

dados = {
    'idade': np.random.normal(35, 10, n).astype(int),
    'plano': np.random.choice(['Gratuito', 'Básico', 'Pro'], n, p=[0.5, 0.3, 0.2]),
    'tempo_uso_horas': np.random.uniform(1, 50, n),
    'bugs_reportados': np.random.poisson(2, n),
    'satisfacao': np.random.randint(1, 11, n)
}

df_sistema = pd.DataFrame(dados)

df_sistema.loc[df_sistema['plano'] == 'Pro', 'tempo_uso_horas'] += 15
df_sistema.loc[df_sistema['plano'] == 'Pro', 'satisfacao'] += 2
df_sistema['satisfacao'] = df_sistema['satisfacao'] - (df_sistema['bugs_reportados'] * 0.5)
df_sistema['satisfacao'] = df_sistema['satisfacao'].clip(1, 10).astype(int)
df_sistema['idade'] = df_sistema['idade'].clip(18, 70)

df_sistema.head()
```

# Parte 1 – Distribuições e Contagens

## 1. Contagem de usuários por plano

**Resposta:** O `countplot` mostra a quantidade de usuários em cada plano. As barras são ordenadas pela frequência, da maior para a menor.


```python
ordem_planos = df_sistema['plano'].value_counts().sort_values(ascending=False).index

plt.figure(figsize=(8, 5))
sns.countplot(data=df_sistema, x='plano', order=ordem_planos)
plt.title('Quantidade de Usuários por Plano')
plt.xlabel('Plano')
plt.ylabel('Quantidade de usuários')
plt.tight_layout()
plt.show()

print(df_sistema['plano'].value_counts())
```

## 2. Distribuição da idade

**Resposta:** O `histplot` apresenta a distribuição das idades e `kde=True` adiciona a curva de densidade para suavizar a visualização.


```python
plt.figure(figsize=(9, 5))
sns.histplot(data=df_sistema, x='idade', bins=20, kde=True)
plt.title('Distribuição da Idade dos Usuários')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.tight_layout()
plt.show()
```

# Parte 2 – Relações Categóricas e Numéricas

## 3. Tempo de uso por plano – Boxplot

**Resposta:** O boxplot compara mediana, dispersão e possíveis valores extremos do tempo de uso entre os planos. Como os dados acrescentam 15 horas ao plano Pro, esse grupo tende a apresentar valores maiores.


```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df_sistema, x='plano', y='tempo_uso_horas')
plt.title('Distribuição do Tempo de Uso por Plano')
plt.xlabel('Plano')
plt.ylabel('Tempo de uso (horas)')
plt.tight_layout()
plt.show()
```

## 4. Satisfação por plano – Violinplot

**Resposta:** O violinplot mostra a distribuição e a densidade das notas de satisfação para cada plano. O Pro tende a apresentar satisfação maior devido ao acréscimo de 2 pontos aplicado na geração dos dados.


```python
plt.figure(figsize=(9, 5))
sns.violinplot(data=df_sistema, x='plano', y='satisfacao')
plt.title('Distribuição da Satisfação por Plano')
plt.xlabel('Plano')
plt.ylabel('Satisfação')
plt.tight_layout()
plt.show()
```

# Parte 3 – Correlações e Análise Multivariada

## 5. Tempo de uso × satisfação

**Resposta:** O `scatterplot` investiga a relação entre tempo de uso e satisfação. `hue` diferencia os planos por cor e `size` varia o tamanho dos pontos conforme o número de bugs reportados.


```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_sistema,
    x='tempo_uso_horas',
    y='satisfacao',
    hue='plano',
    size='bugs_reportados',
    sizes=(30, 180),
    alpha=0.7
)
plt.title('Tempo de Uso × Satisfação')
plt.xlabel('Tempo de uso (horas)')
plt.ylabel('Satisfação')
plt.tight_layout()
plt.show()
```

## 6. Matriz de correlação de Pearson

**Resposta:** A matriz considera apenas as variáveis numéricas. O heatmap usa a paleta divergente `coolwarm` e `annot=True` para mostrar os valores dentro das células.

A correlação varia de -1 a +1: valores positivos indicam associação linear positiva, valores negativos indicam associação linear negativa e valores próximos de zero indicam associação linear fraca.


```python
variaveis_numericas = df_sistema.select_dtypes(include=np.number)
matriz_corr = variaveis_numericas.corr(method='pearson')

plt.figure(figsize=(9, 7))
sns.heatmap(
    matriz_corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    vmin=-1,
    vmax=1
)
plt.title('Matriz de Correlação de Pearson')
plt.tight_layout()
plt.show()

display(matriz_corr)
```

## 7. Bônus – Pairplot

**Resposta:** O `pairplot` permite observar simultaneamente as relações entre tempo de uso, bugs reportados e satisfação. O `hue='plano'` diferencia os usuários de cada plano.


```python
sns.pairplot(
    df_sistema[['tempo_uso_horas', 'bugs_reportados', 'satisfacao', 'plano']],
    hue='plano'
)
plt.show()
```

# Conclusão

As visualizações do Seaborn permitem analisar o perfil de uso dos usuários e comparar os planos de assinatura.

- O `countplot` mostra a distribuição dos usuários por plano.
- O `histplot` mostra a distribuição das idades.
- O `boxplot` permite comparar o tempo de uso entre os planos.
- O `violinplot` mostra a distribuição da satisfação.
- O `scatterplot` relaciona tempo de uso, satisfação, plano e bugs.
- O `heatmap` resume as correlações entre as variáveis numéricas.
- O `pairplot` permite uma análise conjunta das principais variáveis.

Na base sintética, o plano Pro foi configurado para ter maior tempo de uso e maior satisfação, enquanto os bugs reportados reduzem a satisfação. **Esses resultados são padrões criados artificialmente pelos dados da atividade e não devem ser interpretados como conclusões reais sobre clientes.**
